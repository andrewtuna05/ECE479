```python
def getTimeDecayExp(tW, clfLastW=0.5):
    '''
    Newest observation gets weight = 1.
    Oldest observation gets weight = clfLastW.

    To apply an exponential time-decay factor, I replaced the linear weighting function with an exponential function. 
    After sorting observations by time, I computed the cumulative uniqueness weights and normalized them to the interval [0,1].
    This means that the newest obersvation gets weight 1, while the oldest observation gets weight = clfLastW.
    '''
    tW = tW.sort_index()

    # Use cumulative uniqueness as time scale
    cumW = tW.cumsum()

    # Normalize cumulative uniqueness to [0, 1]
    x = (cumW - cumW.iloc[0]) / (cumW.iloc[-1] - cumW.iloc[0])

    # Exponential curve from clfLastW to 1
    clfW = clfLastW ** (1 - x)

    return clfW
```

a) If 2/3 of the labels are 0 and 1/3 are 1, then fitting a classifier without balanced class weights will usually bias the classifier toward predicting 0 

b) Balanced class weights force the classifier to penalize mistakes on class 1 more heavily because class 1 is underrepresented. This makes sense because the goal of the secondary model is not simply just to maximize accuracy, but to correctly identify when the primary model’s signal is worth taking.

c) Before applying balanced class weights, the classifier tends to predict mostly 0s because 0 is the majority class. After applying balanced class weights, the classifier assigns more importance to the minority class 1, so the distribution of predicted labels becomes more balanced, with more predicted 1s. This can improve recall for true positives but may reduce precision or overall accuracy.
