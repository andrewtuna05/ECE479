```python
'''
Original from the textbook
'''
def getBins(events,close):
    #1) prices aligned with events
    events_=events.dropna(subset=['t1'])
    px=events_.index.union(events_['t1'].values).drop_duplicates()
    px=close.reindex(px,method='bfill')
    
    #2) create out object
    out=pd.DataFrame(index=events_.index)
    out['ret']=px.loc[events_['t1'].values].values/px.loc[events_.index]-1
    out['bin']=np.sign(out['ret'])
    return out

'''
    If neither profit-taking nor stop-loss barrier was touched,
    then the vertical barrier was the only one touched, so label = 0.
'''

def updated_getBins(events, close):
    # 1) prices aligned with events
    events_ = events.dropna(subset=['t1'])
    px = events_.index.union(events_['t1'].values).drop_duplicates()
    px = close.reindex(px, method='bfill')

    # 2) create out object
    out = pd.DataFrame(index=events_.index)
    out['ret'] = px.loc[events_['t1'].values].values / px.loc[events_.index].values - 1
    out['bin'] = np.sign(out['ret'])

    # 3) Modification
    if 'pt' in events_.columns and 'sl' in events_.columns:
        vertical_only = events_['pt'].isna() & events_['sl'].isna()
        out.loc[vertical_only, 'bin'] = 0

    return out    
```
