### Chapter 6 Exercises from Del Prado Textbook

### 6.1

Bagging uses random sampling with replacement in order to generate different bootstrap training sets from the same original sample. Because each base learner is trained on a different sample, the fitted models are less correlated. Averaging their predictions reduces forecast variance.

If sampling were done without replacement, bagging could still reduce variance if the subsamples were sufficiently different, but the reduction would generally be weaker. Without replacement, the training sets overlap more systematically and the base learners become more correlated. Since variance reduction from averaging depends on low correlation among the base learners, sampling without replacement makes bagging less effective.

### 6.2

a) Highly overlapping labels make bagging mostly ineffective and can also increase overfitting risk. Bagging relies on bootstrap samples producing sufficiently different base learners. When labels overlap heavily, different training observations are not independent because their outcomes share much of the same return path. Therefore, even different bootstrap samples contain highly redundant information.

As a result, the base learners become strongly correlated, so averaging them does not reduce variance as much. In addition, the model may overfit to repeated or redundant return patterns rather than learning independent predictive structure. The main issue is that the effective sample size is much smaller than the apparent sample size.

b) Out-of-the-bag accuracy is generally not reliable in finance. Even when an observation is out-of-bag, its label may overlap in time with labels used in training. This creates leakage, so out-of-the-bag accuracy is often upward biased.
